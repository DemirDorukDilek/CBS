```
import heapq
from itertools import product

def multi_agent_astar(G, heuristics, start, goal):
    """
    Multi-Agent A* Search

    Args:
        G: networkx.Graph (integer vertices)
        heuristics: [{vertex: dist_to_goal}] index = agent_idx
        start: (agent1_start, agent2_start, ...)
        goal: (agent1_goal, agent2_goal, ...)

    Returns:
        (path, cost) or None
    """
    num_agents = len(start)

    def h(state):
        return sum(heuristics[i][v] for i, v in enumerate(state))

    def valid(curr, nxt):
        if len(set(nxt)) != num_agents:
            return False
        for i in range(num_agents):
            for j in range(i + 1, num_agents):
                if curr[i] == nxt[j] and curr[j] == nxt[i]:
                    return False
        return True

    def successors(state):
        moves = [list(G.neighbors(v)) + [v] for v in state]
        for combo in product(*moves):
            if valid(state, combo):
                yield combo

    counter = 0
    open_set = [(h(start), 0, counter, start, [start])]
    g_scores = {start: 0}
    closed = set()

    while open_set:
        f, g, _, curr, path = heapq.heappop(open_set)

        if curr == goal:
            return path, g

        if curr in closed:
            continue
        closed.add(curr)

        for nxt in successors(curr):
            if nxt in closed:
                continue

            new_g = g + sum(1 for i in range(num_agents) if not (curr[i] == goal[i] and nxt[i] == goal[i]))
            if nxt in g_scores and g_scores[nxt] <= new_g:
                continue

            g_scores[nxt] = new_g
            counter += 1
            heapq.heappush(open_set, (new_g + h(nxt), new_g, counter, nxt, path + [nxt]))

    return None


def reverse_dijkstra(graph, goals):
    rg = graph.reverse()
    h_values = []
    for goal in goals:
        h_values.append(nx.single_source_shortest_path_length(rg,goal))
    return h_values

with open("graph.json","r") as f:
    x = json.load(f)
    mapa,START,END = x["edges"],x["start"],x["goal"]
def rstrip(li):
    to = len(li)
    while to>1 and li[to-1] == li[to-2]:
        to -= 1
    return to
net = nx.DiGraph()
net.add_edges_from(mapa)
START = START[2],START[3],START[4],START[6]
END = END[2],END[3],END[4],END[6]
ee,c = multi_agent_astar(net,reverse_dijkstra(net,END),tuple(START),tuple(END))
x = list(zip(*ee))
for i in x:
    to = rstrip(i)
    k = i[:to]
    print(len(k)-1,k)
```
